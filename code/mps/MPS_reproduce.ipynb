# Complete reproduction of the MPS study

Place this folder beside `GeoEncodeTurb_main.ipynb`. The default mode generates velocity fields on all five grids, 45 MPS representations, spectra, validation results, and the six-panel figure. Matching completed results are reused. Allow 64 GiB of RAM and at least 25 GiB of free disk space for the complete run.

First install the dependencies in `requirements.txt` into the current Python kernel's environment. Modes: `full` runs the complete study; `check` performs a fresh 15- and 18-qubit numerical check; `plot` redraws from saved data; `verify` checks supplied files only.


```python
from pathlib import Path
import subprocess
import sys

# Change MODE to 'check' for a small fresh numerical run.
MODE = 'full'
TEX_MODE = 'auto'  # 'auto', 'required', or 'off'

candidates = [Path.cwd(), Path.cwd() / 'mps']
PACKAGE = next((p for p in candidates if (p / 'run_all.py').is_file() and (p / 'study_config.json').is_file()), None)
if PACKAGE is None:
    raise FileNotFoundError('Start this notebook in the mps folder or its parent project folder.')
print('Python:', sys.executable)
print('Package:', PACKAGE.resolve())
```

```python
options = {'full': [], 'check': ['--check'], 'plot': ['--plot-only'], 'verify': ['--verify-only']}
if MODE not in options:
    raise ValueError('Unknown MODE')
command = [sys.executable, '-u', str(PACKAGE / 'run_all.py'), '--tex', TEX_MODE, *options[MODE]]
print('Running:', command)
subprocess.run(command, check=True, cwd=PACKAGE.parent)
print('Outputs:', PACKAGE / 'generated')
```

## Outputs

Complete results appear in `generated/fields/`, `generated/mps_uniform/`, `generated/recheck/`, and `generated/results/`. The final PDF, PNG, standalone plotting notebook, caption, and two LaTeX tables appear in `generated/figure/`.

The 30-qubit calculation performs parameter compilation and transpilation only, without a corresponding velocity field or MPS. The `plot` mode uses archived reference data; `full` performs all numerical calculations.
