```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from setup_plot import setup_plot

setup_plot()
fontsize = 8
```

```python
df = pd.read_csv(fr'data/3D/turb_PDF_vel.csv', header=0)
vel = np.array(df['vel'].tolist())
pdf = np.array(df['pdf'].tolist())
```

```python
fig_width = 18/2.54
fig_height = 18/2.54
fig = plt.figure(figsize=(fig_width, fig_height))

ax_width = 4/2.54 / fig_width
ax_height = ax_width / 4 * 3

# 
ax = fig.add_axes([0, 0, ax_width, ax_height])
ax.grid(color='k', linestyle='-', linewidth=0.5, alpha=0.1)
ax.plot(vel, pdf, linestyle='none', marker='o', markersize=6, markevery=(0, 0.03), markerfacecolor='#DDE9F5', markeredgecolor='#7AA7D3', markeredgewidth=1)
# ax.plot(val_ux, pdf_ux)
# ax.plot(val_uy, pdf_uy)
# ax.plot(val_uz, pdf_uz)
ax.plot(vel, np.exp(-vel**2), linestyle='--', dashes=(2, 2), color='k', label=r'$\mathrm{e}^{-A|u_j/u^\prime|}$')
# plt.axvline(x=0, color='k', linestyle='--', dashes=(2, 2))
plt.xlim(-3, 3)
plt.ylim(1e-12, 1e2)
ax.legend(ncol=1, frameon=True, labelspacing=0.2, handlelength=2.5, handletextpad=0.5, bbox_to_anchor=(0.5, 0), loc='lower center', fontsize=fontsize-1)
plt.yscale('log')
plt.xlabel(r'$u_j/u^\prime$', labelpad=0, size=fontsize)
plt.ylabel(r'PDF', labelpad=1, size=fontsize) # rotation=0
ax.tick_params(axis='x', which='major', direction='out', top=False, right=False, length=3, pad=1.5)
ax.tick_params(axis='y', which='major', direction='out', top=False, right=False, length=3, pad=1)
ax.tick_params(which='minor', direction='out', top=False, right=False, length=1.5)
```
